```python
def amount():
  price=float(input("Please input how much money the items were: "))
  gave=float(input("Please input how much you gave: "))
  if price==gave:
    print("You are good to go")
  elif price>gave:
    print("You still have money to pay")
  elif gave>price:
    return gave-price
  balance=amount()
  print(f"your balance is {balance}" )
amount()
```

```python
num=int(input("Please input any number"))
t=num
numLen=0
while t>0:
  numLen+=1
  t=t//10

if numLen>=4:
  target_digit_position = int(numLen/2)
  chk=0
  midOne=0
  midTwo=0
  temp_num = num
  while temp_num > 0:
    rem=temp_num%10

    if chk==target_digit_position:
      midOne=rem
    elif chk==(target_digit_position-1):
     midTwo=rem
    temp_num=temp_num//10
    chk+=1
  prod=midOne*midTwo
  print(f"Product of mid digits {midOne}*{midTwo}={prod}")
else:
  print("It is not a 4 or more than 4-digit number")
```

```python
rows=int(input("Please input the number of rows you want in your mirrored right angle triangle"))
print("Mirrored right angle triangle")
for i in range(1,rows+1):
  for j in range(1,(rows-i)+1):
    print(" ",end=" ")
  for k in range(i):
    print("*",end =" ")
  print()
```

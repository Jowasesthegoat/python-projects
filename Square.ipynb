```python
from ColabTurtlePlus.Turtle import *
initializeTurtle()
bgcolor("red")
setup(400,500)
square=Turtle()
n_sides=4
side_length=50
angle=360/n_sides
for i in range(n_sides):
  square.forward(side_length)
  square.right(angle)
done()
```

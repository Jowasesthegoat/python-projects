```python
start_num=int(input("Please input the first number"))
end_num=int(input("Please input the last number"))
even=[]
odd=[]
for num in range(start_num, end_num+1):
  sqaure_value=num**2
  if sqaure_value % 2 == 0:
    even.append(sqaure_value)
  else:
    odd.append(sqaure_value)
print("The even numbers are",even)
print("The odd numbers are",odd)
```
